1. Why Pipelines are important in ML?

A pipeline allows us to combine multiple ML steps into one sequential workflow. It help to create a chain for the steps involved in preprocessing of the raw data.
This prevent Data Leakage.
Pipelines also make code look cleaner and we can define the workflowe once.

2. What Problems do Pipeline Solve?

The biggest problem is:How do we ensure that all preprocessing and model-training steps happen correctly, and without data leakage?
Pipeline ensure that preprocessing operations such as scaling and encoding are fitted only on the appropriate training data during cross-validation.
By combining multiple steps into one manageable workflow, they make complex code look cleaner.


3. Difference Between Manual Processing and Pipeline-based Processing.

Manual processing means performing each preprocessing and machine learning step separately. The developer has to manually ensure that every transformation is fitted correctly and applied consistently.

Pipeline- Based processing combines these steps into a single sequence, where each step is automatically executed in the defined order.

| Manual Processing                                  | Pipeline Processing                       |
| -------------------------------------------------- | ----------------------------------------- |
| Data Leakage.                                      |No data Leakage.                           |
 Each step is performed separately.                 | All steps are combined into one workflow. |
| More chances of human error.                       | Reduces chances of errors.                |
| Preprocessing must be managed manually.            | Preprocessing is handled automatically.   |
| Greater risk of inconsistent preprocessing.        | Maintains consistency.                    |
