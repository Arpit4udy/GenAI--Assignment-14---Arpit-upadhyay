```python
import pandas as pd
```

```python
df=pd.read_csv('dataset.csv')
```

```python
df.columns
```

```python
numerical=['customer_id', 'age', 'annual_income', 'months_active',
       'avg_monthly_spend', 'purchase_frequency', 'avg_order_value',
       'discount_usage_rate', 'return_rate', 'browsing_time_minutes',
       'support_interactions','total_spend', 'engagement_score', 'total_spend.1',
       'engagement_score.1']
categorical=['payment_method', 'region']
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
```

```python
numerical_pipeline=Pipeline(steps=[('numerical',StandardScaler())])
categorical_pipeline=Pipeline(steps=[('encode',OneHotEncoder(handle_unknown='ignore'))])
```

```python
from sklearn.compose import ColumnTransformer
```

```python
features=ColumnTransformer(transformers=[('num',numerical_pipeline,numerical),('cat',categorical_pipeline,categorical)])
```

```python
features
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=Pipeline(steps=[('features',features),('Classifier',LogisticRegression())])
```

```python
model
```

```python
from sklearn.model_selection import train_test_split
```

```python
X=df.drop(columns=['customer_segment'])
Y=df['customer_segment']
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
Y=le.fit_transform(Y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2)
```

```python
model.fit(x_train,y_train)
```

```python
y_predict=model.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_predict)*100
```
